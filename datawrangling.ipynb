```python
import pandas as pd
df = pd.read_csv('https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/HLOosvsPgIwt5dgOOh1RSg/survey-data-updated.csv')
df = df[['LanguageHaveWorkedWith', 'LanguageWantToWorkWith', 'DatabaseHaveWorkedWith', 'DatabaseWantToWorkWith', 'PlatformHaveWorkedWith', 'PlatformWantToWorkWith', 'WebframeHaveWorkedWith', 'WebframeWantToWorkWith','Age','Country','EdLevel']]
df.shape
```

```python
def explode_and_count(df, col, label):
    tmp = df[[col]].dropna().copy()
    tmp[col] = tmp[col].str.split(';')
    tmp = tmp.explode(col)
    tmp[col] = tmp[col].str.strip()

    out = (
        tmp[col]
        .value_counts()
        .reset_index()
    )
    out.columns = ['Technology', 'Count']
    out['Category'] = label
    return out

lang_used = explode_and_count(df, 'LanguageHaveWorkedWith', 'Language Used')
db_used   = explode_and_count(df, 'DatabaseHaveWorkedWith', 'Database Used')
plat_used = explode_and_count(df, 'PlatformHaveWorkedWith', 'Platform Used')
web_used  = explode_and_count(df, 'WebframeHaveWorkedWith', 'Web Framework Used')

lang_want = explode_and_count(df, 'LanguageWantToWorkWith', 'Language Desired')
db_want   = explode_and_count(df, 'DatabaseWantToWorkWith', 'Database Desired')
plat_want = explode_and_count(df, 'PlatformWantToWorkWith', 'Platform Desired')
web_want  = explode_and_count(df, 'WebframeWantToWorkWith', 'Web Framework Desired')

final_df = pd.concat([
    lang_used,
    db_used,
    plat_used,
    web_used,
    lang_want,
    db_want,
    plat_want,
    web_want
], ignore_index=True)

#final_df.to_csv("survey_results_looker_ready.csv", index=False)
final_df
```
